```python
from os import read
import pandas as pd

file = pd.read_csv("entropy_analysis/llma31_8b/longbench/hotpotqa/tf_decode/records.csv")

matrix = pd.DataFrame(columns=["position", "75-25"])
for i in file["position"]:
    print(i)
  
    mask_25 = (file["ratio"] == 0.25) & (file["position"] == i)
    mask_75 = (file["ratio"] == 0.75) & (file["position"] == i)

    h_compressed_25 = file.loc[mask_25, "h_compressed"].sum()
    h_compressed_75 = file.loc[mask_75, "h_compressed"].sum()
    print(h_compressed_25, h_compressed_75)
    h_compressed_diff = h_compressed_75 - h_compressed_25
    print(h_compressed_diff)
    matrix = pd.concat([matrix, pd.DataFrame({"position": [i], "75-25": [h_compressed_diff]})], ignore_index=True)
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.figure(figsize=(10, 6))
plt.hist(matrix["75-25"], bins=20, edgecolor='black')
plt.title("Histogram of Differences in Compressed Entropy (75% - 25%)")
plt.xlabel("Difference in Compressed Entropy")
plt.ylabel("Frequency")
plt.grid(axis='y', alpha=0.75)
plt.savefig("entropy_analysis/llma31_8b/longbench/hotpotqa/tf_decode/histogram_75_25.png")
plt.show()
```
